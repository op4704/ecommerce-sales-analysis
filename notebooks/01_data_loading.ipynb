## Data dictionary

| Column | Meaning |
|---|---|
| InvoiceNo | Invoice/transaction ID. A 'C' prefix means the invoice was cancelled. |
| StockCode | Product (item) code. |
| Description | Product name. |
| Quantity | Units sold in this transaction line. Negative = return. |
| InvoiceDate | Date and time of the transaction. |
| Price | Unit price in GBP. |
| Customer ID | Unique customer identifier. Many rows have this missing (guest/unlinked orders). |
| Country | Customer's country. |

Knowing this up front tells us *what* to check for during cleaning:
cancelled invoices, negative quantities, missing customer IDs, and price
anomalies (zero or negative prices).
